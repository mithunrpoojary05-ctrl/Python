```python
total_sales = df["Sales"].sum()
total_expenses = df["Expenses"].sum()
total_profit = df["Profit"].sum()

average_sales = df["Sales"].mean()
highest_sales = df["Sales"].max()

print("Total Sales:", total_sales)
print("Total Expenses:", total_expenses)
print("Total Profit:", total_profit)
print("Average Sales:", average_sales)
print("Highest Sales:", highest_sales)
```

```python
monthly=df.groupby("Month")[["Sales","Expenses","Profit"]].sum()
print("\n========Monthly summary========")
print(monthly)
```

```python
import seaborn as sns
```

```python
import matplotlib.pyplot as plt
```

```python
monthly=monthly.sort_values(by="Sales")
```

```python
# Line chart with plt
plt.figure(figsize=(12,6))
plt.plot(monthly["Sales"], monthly["Profit"], label="monthly trend", marker='o')
plt.title("sales and profit of monthly expenses")
plt.xlabel("sales")
plt.ylabel("profit")
plt.grid(True)
plt.legend()
plt.xticks(rotation=45)
plt.show()
```

```python
import seaborn as sns
```

```python
#bar chart
plt.figure(figsize=(12, 6))
sns.barplot( 
    x='Month',      
    y='Sales', 
    data=monthly, 
)
plt.title("Monthly Sales and Profit Comparison")
plt.xlabel("Month")
plt.ylabel("sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#scatter plot
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Sales', y='Profit',data=monthly,hue="Sales", color='coolwarm')
plt.title('Sales vs. Profit')
plt.xlabel('Sales')
plt.ylabel('Profit')
plt.grid(True)
plt.xticks(rotation=45)
plt.show()
```

```python
#profit margin
monthly["profit margin"]=(monthly['Profit']/monthly['Sales'])*100
monthly
```

```python
numeric_cols=["Sales","Expenses","Profit","Assets","Liabilities","Stock_Price"]
monthly=df[numeric_cols].corr()
monthly
```

```python
#heat map
sns.heatmap(monthly,
           annot=True,
           cmap="coolwarm",
           cbar=False,
           square=True,
           linewidth=1)
plt.title("profit_margin")
plt.show()
```
